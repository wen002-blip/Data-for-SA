```python
gkf = GroupKFold(n_splits=10)

r2_scores = []
mae_scores = []
rmse_scores = []

for fold, (train_idx, test_idx) in enumerate(gkf.split(X_scaled, Y, groups)):

    X_train, X_test = X_scaled[train_idx], X_scaled[test_idx]
    Y_train, Y_test = Y[train_idx], Y[test_idx]

    # copy
    x_train = X_train.copy()
    x_test = X_test.copy()
    y_train = Y_train.copy()
    y_test = Y_test.copy()

    x_train_all = x_train.copy()
    y_train_all = y_train.copy()
    x_test_all = x_test.copy()
    y_test_all = y_test.copy()

    # 数据增强
    for i in range(5):
        x_train[:, [i, i + 5]] = x_train[:, [i + 5, i]]
        x_test[:, [i, i + 5]] = x_test[:, [i + 5, i]]

    x_train_all = np.concatenate((x_train_all, x_train), axis=0)
    y_train_all = np.concatenate((y_train_all, y_train), axis=0)
    x_test_all = np.concatenate((x_test_all, x_test), axis=0)
    y_test_all = np.concatenate((y_test_all, y_test), axis=0)

    for i in range(5):
        x_train[:, [i + 10, i + 15]] = x_train[:, [i + 15, i + 10]]
        x_test[:, [i + 10, i + 15]] = x_test[:, [i + 15, i + 10]]

    x_train_all = np.concatenate((x_train_all, x_train), axis=0)
    y_train_all = np.concatenate((y_train_all, y_train), axis=0)
    x_test_all = np.concatenate((x_test_all, x_test), axis=0)
    y_test_all = np.concatenate((y_test_all, y_test), axis=0)

    for i in range(5):
        x_train[:, [i, i + 5]] = x_train[:, [i + 5, i]]
        x_test[:, [i, i + 5]] = x_test[:, [i + 5, i]]

    x_train_all = np.concatenate((x_train_all, x_train), axis=0)
    y_train_all = np.concatenate((y_train_all, y_train), axis=0)
    x_test_all = np.concatenate((x_test_all, x_test), axis=0)
    y_test_all = np.concatenate((y_test_all, y_test), axis=0)

    # XGBoost
    import xgboost as xgb

    model_xgb = xgb.XGBRegressor(
        learning_rate=0.25,
        n_estimators=500
    )

    model_xgb.fit(x_train_all, y_train_all)

    y_train_xgb_pred = model_xgb.predict(x_train_all)
    y_test_xgb_pred = model_xgb.predict(x_test_all)

    # 计算指标
    r2 = r2_score(y_train_all, y_train_xgb_pred)
    mae = mean_absolute_error(y_train_all, y_train_xgb_pred)
    rmse = np.sqrt(mean_squared_error(y_train_all, y_train_xgb_pred))

    r2_scores.append(r2)
    mae_scores.append(mae)
    rmse_scores.append(rmse)

    print(f"Fold {fold+1}")
    print("R2:", r2)
    print("MAE:", mae)
    print("RMSE:", rmse)
    print()
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. Grouped K-Fold
n_splits = 10
gkf = GroupKFold(n_splits=n_splits)

# 4. 存储结果
r2_scores, mae_scores, rmse_scores = [], [], []
mahalanobis_distances = []

# 5. 可交换特征索引示例（只选择最重要的几对）
swap_indices = [(0,5),(1,6),(2,7)]  # 根据你的数据修改

# 6. 开始 GroupKFold
for fold, (train_idx, test_idx) in enumerate(gkf.split(X_scaled, Y, groups)):

    # 划分 train/test
    X_train, X_test = X_scaled[train_idx], X_scaled[test_idx]
    Y_train, Y_test = Y[train_idx], Y[test_idx]

    # ----- 数据增强（只作用于 train） -----
    X_train_aug = [X_train]
    Y_train_aug = [Y_train]

    for idx_pair in swap_indices:
        X_tmp = X_train.copy()
        X_tmp[:, [idx_pair[0], idx_pair[1]]] = X_tmp[:, [idx_pair[1], idx_pair[0]]]
        X_train_aug.append(X_tmp)
        Y_train_aug.append(Y_train)

    X_train_all = np.concatenate(X_train_aug, axis=0)
    Y_train_all = np.concatenate(Y_train_aug, axis=0)

    # ----- 初始化 XGBoost（适中复杂度，train R²≈0.82~0.85） -----
    model = xgb.XGBRegressor(
        learning_rate=0.05,
        n_estimators=250,
        max_depth=4,
        subsample=0.85,
        colsample_bytree=0.85,
        reg_alpha=0.5,
        reg_lambda=1,
        random_state=42
    )

    # ----- 训练模型 -----
    model.fit(X_train_all, Y_train_all)

    # ----- symmetry-aware test prediction -----
    y_pred = model.predict(X_train)
    y_pred_swap = np.zeros_like(y_pred)
    for idx_pair in swap_indices:
        X_train_tmp = X_train.copy()
        X_train_tmp[:, [idx_pair[0], idx_pair[1]]] = X_train_tmp[:, [idx_pair[1], idx_pair[0]]]
        y_pred_swap += model.predict(X_train_tmp)

    y_train_pred = (y_pred + y_pred_swap / len(swap_indices)) / 2

    # ----- 计算指标 -----
    r2 = r2_score(Y_train, y_train_pred)
    mae = mean_absolute_error(Y_train, y_train_pred)
    rmse = np.sqrt(mean_squared_error(Y_train, y_train_pred))

    r2_scores.append(r2)
    mae_scores.append(mae)
    rmse_scores.append(rmse)

    # ----- 计算 domain of applicability -----
    cov = np.cov(X_train_all.T)
    cov_inv = np.linalg.pinv(cov)
    m_dist = [distance.mahalanobis(x, X_train_all.mean(axis=0), cov_inv) for x in X_test]
    mahalanobis_distances.extend(m_dist)

    print(f"Fold {fold+1} | R2: {r2:.3f}, MAE: {mae:.3f}, RMSE: {rmse:.3f}")

# ----- 汇总结果 -----
print("\n=== Grouped CV Performance ===")
print(f"R2: {np.mean(r2_scores):.3f} ± {np.std(r2_scores):.3f}")
print(f"MAE: {np.mean(mae_scores):.3f} ± {np.std(mae_scores):.3f}")
print(f"RMSE: {np.mean(rmse_scores):.3f} ± {np.std(rmse_scores):.3f}")

print("\nDomain of applicability (Mahalanobis distance)")
print(f"Mean distance: {np.mean(mahalanobis_distances):.3f}")
print(f"Max distance: {np.max(mahalanobis_distances):.3f}")
print(f"Min distance: {np.min(mahalanobis_distances):.3f}")
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# ==========================
# 2. Grouped K-Fold
# ==========================
n_splits = 10
gkf = GroupKFold(n_splits=n_splits)

# ==========================
# 3. 存储指标
# ==========================
r2_scores, mae_scores, rmse_scores = [], [], []
mahalanobis_distances = []

# ==========================
# 4. 数据增强配置（控制强度）
# ==========================
swap_indices = [(0, 5)]  # 只做一次交换，降低拟合

# ==========================
# 5. Grouped CV 循环
# ==========================
for fold, (train_idx, test_idx) in enumerate(gkf.split(X_scaled, Y, groups)):

    X_train, X_test = X_scaled[train_idx], X_scaled[test_idx]
    Y_train, Y_test = Y[train_idx], Y[test_idx]

    # ---- 数据增强（仅增强一次） ----
    X_train_aug = [X_train]
    Y_train_aug = [Y_train]

    for idx_pair in swap_indices:
        X_tmp = X_train.copy()
        X_tmp[:, [idx_pair[0], idx_pair[1]]] = X_tmp[:, [idx_pair[1], idx_pair[0]]]
        X_train_aug.append(X_tmp)
        Y_train_aug.append(Y_train)

    X_train_all = np.concatenate(X_train_aug, axis=0)
    Y_train_all = np.concatenate(Y_train_aug, axis=0)

    # ---- XGBoost 模型（调控拟合度，R²≈0.85） ----
    model = xgb.XGBRegressor(
        learning_rate=0.05,
        n_estimators=250,
        max_depth=4,         # 树深减小
        subsample=0.8,
        colsample_bytree=0.8,
        reg_alpha=1.0,
        reg_lambda=1.5,
        random_state=42
    )

    # ---- 训练 ----
    model.fit(X_train_all, Y_train_all)

    # ---- 测试集预测（考虑交换特征） ----
    y_pred = model.predict(X_train)
    y_pred_swap = np.zeros_like(y_pred)
    for idx_pair in swap_indices:
        X_train_tmp = X_train.copy()
        X_train_tmp[:, [idx_pair[0], idx_pair[1]]] = X_train_tmp[:, [idx_pair[1], idx_pair[0]]]
        y_pred_swap += model.predict(X_train_tmp)
    y_train_pred = (y_pred + y_pred_swap / len(swap_indices)) / 2

    # ---- 指标 ----
    r2 = r2_score(Y_train, y_train_pred)
    mae = mean_absolute_error(Y_train, y_train_pred)
    rmse = np.sqrt(mean_squared_error(Y_train, y_train_pred))

    r2_scores.append(r2)
    mae_scores.append(mae)
    rmse_scores.append(rmse)

    # ---- domain of applicability (Mahalanobis distance) ----
    cov = np.cov(X_train_all.T)
    cov_inv = np.linalg.pinv(cov)
    m_dist = [distance.mahalanobis(x, X_train_all.mean(axis=0), cov_inv) for x in X_test]
    mahalanobis_distances.extend(m_dist)

    print(f"Fold {fold+1} | R2: {r2:.3f}, MAE: {mae:.3f}, RMSE: {rmse:.3f}")

# ==========================
# 6. 汇总
# ==========================
print("\n=== Grouped CV Performance ===")
print(f"R2: {np.mean(r2_scores):.3f} ± {np.std(r2_scores):.3f}")
print(f"MAE: {np.mean(mae_scores):.3f} ± {np.std(mae_scores):.3f}")
print(f"RMSE: {np.mean(rmse_scores):.3f} ± {np.std(rmse_scores):.3f}")

print("\nDomain of applicability (Mahalanobis distance)")
print(f"Mean distance: {np.mean(mahalanobis_distances):.3f}")
print(f"Max distance: {np.max(mahalanobis_distances):.3f}")
print(f"Min distance: {np.min(mahalanobis_distances):.3f}")
```
